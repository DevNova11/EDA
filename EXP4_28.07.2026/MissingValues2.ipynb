# Imputation

```
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer

new_df = pd.DataFrame(
    {
        "Score": [85, 92, np.nan, 78, 65, 90, np.nan, 88, 72, 80],
        "Hours_Studied": [5, np.nan, 7, 4, 3, 6, 8, np.nan, 3, 5],
        "Grade": ["A", "A", "B", "C", "D", "A", "B", "A", "C", np.nan],
    }
)

mean_imputer_new = SimpleImputer(strategy="mean")
new_df["Score_Mean"] = mean_imputer_new.fit_transform(new_df[["Score"]]).ravel()

median_imputer_new = SimpleImputer(strategy="median")
new_df["Hours_Studied_Median"] = median_imputer_new.fit_transform(
    new_df[["Hours_Studied"]]
).ravel()

mode_imputer_new = SimpleImputer(strategy="most_frequent")
new_df["Grade_Mode"] = mode_imputer_new.fit_transform(new_df[["Grade"]]).ravel()

num_cols_new = ["Score", "Hours_Studied"]
mice_imputer_new = IterativeImputer(max_iter=10, random_state=42)
df_mice_new = new_df[num_cols_new].copy()
df_mice_new.iloc[:, :] = mice_imputer_new.fit_transform(df_mice_new)
new_df["Score_MICE"] = df_mice_new["Score"]
new_df["Hours_Studied_MICE"] = df_mice_new["Hours_Studied"]

print(new_df)
```

# Missing Values

```
import numpy as np
import pandas as pd

# 1. Create a DataFrame with initial and added missing values
print("### 1. Creating DataFrame with missing values")
sales_data = np.random.randint(100, 1000, size=(5, 3))
sales_df = pd.DataFrame(sales_data,
                        index=['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
                        columns=['Region_A', 'Region_B', 'Region_C'])

sales_df['Region_D'] = np.nan # Entire column NaN initially
sales_df.loc['Tablet'] = np.random.randint(50, 500, size=4)
sales_df.loc['Speaker'] = np.nan # Entire row NaN
sales_df.loc['Mouse', 'Region_A'] = np.nan
sales_df.loc['Keyboard', 'Region_C'] = np.nan
sales_df.loc['Monitor', 'Region_D'] = 250.0 # Fill one NaN
display(sales_df)
print("\n")

# 2. Inspecting NaN values
print("### 2. Inspecting NaN values")
print("NaN count per column:")
display(sales_df.isnull().sum())
print("Total NaNs:", sales_df.isnull().sum().sum())
print("\n")

# 3. Dropping missing values
print("### 3. Dropping missing values")
print("Original:")
display(sales_df)
print("\nDrop rows with ANY NaN:")
display(sales_df.dropna())
print("\nDrop rows where ALL NaN:")
display(sales_df.dropna(how='all'))
print("\nDrop cols where ALL NaN:")
display(sales_df.dropna(how='all', axis=1))
print("\nDrop cols with < 5 non-NaNs:")
display(sales_df.dropna(thresh=5, axis=1))
print("\n")

# 4. Mathematical operations with NaN
print("### 4. Mathematical operations with NaN")
ar_sales = np.array([100, 200, np.nan, 300, 150])
ser_sales = pd.Series(ar_sales)
print(f"Series: {ser_sales.tolist()}")
print(f"Mean (ignores NaN): {ser_sales.mean()}")
print(f"Sum of 'Region_D': {sales_df['Region_D'].sum()}")
print(f"Cumsum of 'Region_A':\n{sales_df['Region_A'].cumsum().to_string()}")
print("\n")

# 5. Filling missing values
print("### 5. Filling missing values")
print("Original:")
display(sales_df)
print("\nFill NaN with 0:")
display(sales_df.fillna(0))
print("\nForward-fill 'Region_A':")
display(sales_df['Region_A'].fillna(method='ffill'))
print("\nBackward-fill 'Region_A':")
display(sales_df['Region_A'].fillna(method='bfill'))
```
