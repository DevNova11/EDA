## Code 1: Fibonacci Series

```python
n = int(input("Enter the number of terms: "))
a, b = 0, 1

for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b

print()
```

## Code 2: Factorial

```python
n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)
```

## Code 3: Prime Number Checker

```python
n = int(input("Enter a number: "))

if n > 1:
    for i in range(2, n):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
else:
    print("Not Prime")
```

## Code 4: Palindrome Checker

```python
text = input("Enter a word: ")

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

## Code 5: Armstrong Number Checker

```python
n = int(input("Enter a number: "))

temp = n
digits = len(str(n))
total = 0

while temp > 0:
    digit = temp % 10
    total += digit ** digits
    temp //= 10

if total == n:
    print("Armstrong Number")
else:
    print("Not Armstrong")
```

## Code 6: Reverse a Number

```python
n = int(input("Enter a number: "))

reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10

print("Reversed Number:", reverse)
```

## Code 7: Sum of Digits

```python
n = int(input("Enter a number: "))

total = 0

while n > 0:
    total += n % 10
    n //= 10

print("Sum of digits =", total)
```

## Code 8: Multiplication Table

```python
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

## Code 9: Even or Odd Checker

```python
n = int(input("Enter a number: "))

if n % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## Code 10: Number Guessing Game

```python
import random

secret = random.randint(1, 10)
guess = int(input("Guess a number (1-10): "))

if guess == secret:
    print("Correct!")
else:
    print("Wrong! The number was", secret)
```
